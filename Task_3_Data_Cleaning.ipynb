## 2. Missing Data Handling Strategy

The dataset contains missing values in Age, Cabin, and Embarked.

- Age: Missing values will be replaced with the median age because age is
  a numerical variable and the median is less affected by extreme values.
- Cabin: Missing values will be replaced with "Unknown" because Cabin has
  a large number of missing values, and deleting the rows would remove a
  substantial portion of the dataset.
- Embarked: Missing values will be replaced with the mode because Embarked
  is a categorical variable and only a small number of values are missing.

This approach preserves the maximum amount of useful data while using
appropriate methods for each variable.

```python
age_median = df["Age"].median()

age_median
```

```python
df["Age"] = df["Age"].fillna(age_median)
```

```python
df["Age"].isnull().sum()
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    return Q1, Q3, IQR, lower_bound, upper_bound, len(outliers)
```

```python
age_result = detect_outliers_iqr(df, "Age")

age_result
```

fare_result = detect_outliers_iqr(df, "Fare")

## 4. Outlier Detection

The IQR method was used to identify potential outliers in the numerical
variables Age and Fare.

The IQR method defines potential outliers as observations below
Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR.

Potential outliers were not automatically removed. Extreme values were
reviewed to determine whether they represented invalid data or genuine
observations.

Age values were retained when they were within a realistic human age
range. High Fare values were also retained because an expensive ticket
can represent a legitimate transaction rather than a data-entry error.

Therefore, statistically identified but plausible observations were
retained instead of being unnecessarily removed.
